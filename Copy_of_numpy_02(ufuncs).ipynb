**ufuncs ** *(universal function)*
vectoorization--process of converting iterative statement into vector -b

Vectorization means:

Converting an iterative (loop-based) operation into an array operation.

```python
x=[1,2,3,4]
y = [4,5,6,7]
z = np.add(x,y)
print(z)
```

Here:

myadd (first argument) → your Python function
2 → takes 2 inputs (x and y)
1 → returns 1 output

np.frompyfunc() converts your normal Python function into a NumPy ufunc.

```python

# create it own ufunc -- using frompfunc() method

def myadd(x,y):
  return x+y

myadd = np.frompyfunc(myadd,2,1)   # frompyfunc command is use to customize your own ufuncs
print(myadd(y,z))
print(myadd([1,2,3,4],[5,6,7,8]))
```

# ***CONCATENATE FUNCTION - The np.concatenate() function is used to join two or more arrays into a single array***

```python
arr1=np.array([1,2,3])
arr2=np.array([42,52,62])
arr=np.concatenate((arr1,arr2))   # make sure used ( ) pranenthesis  brackets rather then [ ]
print(arr)
```

```python
# join the 2-D array along axis
arr1=np.array([[10,20,30],[4,5,6]])
arr2=np.array([[70,85,97],[10,11,12]])
arr=np.concatenate((arr1,arr2),axis=1)
print(arr)
```

```python
arr1=np.array([[10,20,30],[4,5,6]])
arr2=np.array([[70,85,97],[10,11,12]])
arr=np.concatenate((arr1,arr2),axis=0)
print(arr)
```

**type of stack**

1.   hstack() : it will concatenate along the rows
2.   vstack() : it will conactenate along the colm
3.   dstack() : it will concatenate along the deepth


```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.hstack((arr1,arr2))
print(arr)
```

**vstack**

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.vstack((arr1,arr2))
print(arr)
```

**dstack**

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.dstack((arr1,arr2))
print(arr)
```

```python
# the indexes that get a match,to search a array we use where() function

arr = np.array([1,2,3,4,5,4,4])
x = np.where(arr == 4)
print(x)
```

```python
arr = np.array([1,2,3,3,4,5,4,4])
x = np.where((arr == 4) | (arr==3))
print(x)
```

```python
arr = np.array([10,14,93,41,8,7])
x = np.where(arr % 2==0) # even
m = np.where(arr%2==1)
print(m)
print(x)
```

```python
arr = np.array([6,7,8,9])
x = np.searchsorted(arr,7)
print(x)
```
