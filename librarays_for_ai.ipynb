```python
import matplotlib.pyplot as plt            #Graphs aur charts banane ke liye
import numpy as np      #Numerical calculations aur arrays ke liye
import seaborn as sns  #Beautiful aur advanced statistical graphs ke liye
import pandas as pd    #Data ko read, clean aur analyze karne ke liye
```

```python
a = [1,2,3,4,5]
b=[10,12,13,16,17]

# seaborn and matplotlib are connected ,to make a graph we use seaborn while plt.title(), plt.xlabel(), plt.ylabel() this command ar of matplotib

plt.plot(a,b)

plt.title("Line Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.grid(True)

plt.show()
```

```python
categories= ["aaple","Lenovo","acer"]
value =[100,200,300]
plt.bar(categories,value,color=["red","green","pink"],hatch="*")
plt.title("Bar Plot")
plt.xlabel("Categoreies")
plt.ylabel("Values")
plt.show()
```

```python
study_hours = [1,2,3,4,5]
exam_scores =[10,12,13,16,17]

plt.scatter(study_hours,exam_scores,color="red",marker="*")  # if we are not use marker then there is by deafault point() [plt.Scatter  use for ploting]
plt.title("Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")

plt.show()
```

```python
#sub plot (to utilze the space)

fig,axs = plt.subplots(1,4,figsize=(10,4))
axs[0].bar(["Del","Mum","HR"],[3,5,7])
axs[0].set_title("Bar")

axs[1].plot([1,2,3],[4,5,6])
axs[1].set_title("line")

axs[2].pie([3, 5, 7],
    labels=["Delhi", "Mumbai", "Haryana"],
    autopct="%1.1f%%")
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 5]

axs[3].fill_between(x, y)
axs[3].set_title("Area Plot")

plt.tight_layout()
plt.show()
```

```python
days = ["mon","tue","web","thu"]
size = [20,25,30,35]
plt.figure(figsize=(6,6))
plt.pie(size,labels=days,autopct="%1.1f%%",shadow=True,explode=[0.2,0.0,0.0,0.0])
plt.title("Days Scale")
plt.show()
```

```python
month = ["jan","feb","mar","apr","may","June",
         "july","aug","sep","oct","nov","dec"]
sales_2024 =[50,55,53,60,70,75,72,68,70,74,80,74]
sales_2025 =[]
for i in sales_2024:
  sales_2025.append(i+10)
plt.plot(month,sales_2024,label="Sale 2024",color="blue",marker="o")
plt.plot(month,sales_2025,label="Sale 2025",color="red",marker="s")
plt.xlabel("mppnth")
plt.ylabel("sales (in thousand)")
plt.title("monthly sales Trend:2024 vs 2025")
plt.legend()
plt.show()
```
