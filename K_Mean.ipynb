```python
import sklearn
from sklearn.cluster import KMeans
import pandas as pd
import numpy as np

# Data
data = np.array([
    [20, 80],
    [25, 85],
    [30, 90],
    [100, 20],
    [110, 25],
    [120, 30]
])

# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42)
clusters = kmeans.fit_predict(data)

# Print clusters
print("Cluster Assignment:",clusters)
print("cluster cennters:",Kmeans.cluster_centers_)
```
