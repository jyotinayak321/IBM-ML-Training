errors='coerce' ka matlab:*Agar koi value number mein convert nahi ho sakti, usko NaN bana do.*

```python
# i remove the $ Kyunki $2.50 Pandas ke liye text/string hai, number nahi.

df['tip'] = df['tip'].str.replace('$', '', regex=False)
df['tip'] = pd.to_numeric(df['tip'], errors='coerce')   #Text ko number mein convert kiya

x = df['tip'].mean()

df.fillna({'tip': x}, inplace=True)

print(df.head(13))
```
