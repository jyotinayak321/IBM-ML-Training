# **Working with Dates**

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

# **1. Extract year**

```python
#date se year extract karke ek naya column bana
date['date_year'] = date['date'].dt.year

date.sample(5)
```

# **2. Extract Month**

```python
date['date_month_no'] = date['date'].dt.month

date.head()
```

# **Extract Month Name**



```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

# **Extract Days**

```python
date['date_day'] = date['date'].dt.day

date.head()
```

# ***# day of week***

```python
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

# **# day of week - name**

```python
date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head(10)
```

## **is weekend?**

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)  #isin check where the sunday and saturday is present aur not

date.drop(columns=['product_id','city_id','orders']).head()
```

# **Extract week of the year**

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.drop(columns=['product_id','city_id','orders']).head()
```

# **Extract Quarter**

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

# **Extract Semester**

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

# **Extract Time elapsed between dates**




```python
import datetime

today = datetime.datetime.today()       #Current date + time return karta hai.

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days   # diff b/w the current date and the dataframe date
```

```python

# Months passed
np.round((today - date['date']).dt.days / 30, 0)
```
