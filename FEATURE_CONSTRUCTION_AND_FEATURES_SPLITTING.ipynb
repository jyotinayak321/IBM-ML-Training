**Feacture Construction** is the process of manually creating new features from existing features by combining, transforming or deriving them based on domain knowledge and intuition.


```python
import pandas as pd
df = pd.read_csv("weight-height (1).csv")
```

```python
# New features
df['Height_m'] = df['Height'] * 0.0254
df['Weight_kg'] = df['Weight'] * 0.453592

df.head()
```

```python
df['BMI'] = df['Weight_kg'] / (df['Height_m'] ** 2)

df.head()
```

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal'
    elif bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'

df['BMI_Category'] = df['BMI'].apply(bmi_category)

df.head()
```

**Feature Splitting ** is the process of breaking down a single columns that contain multiple pieces of information into separate ,atomic columns.this adresses non tidy data where multiple value are stored in a single cell.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Height_scaled', 'Weight_scaled']] = scaler.fit_transform(
    df[['Height', 'Weight']]
)

df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Height_scaled', 'Weight_scaled']] = scaler.fit_transform(
    df[['Height', 'Weight']]
)

df.head()
```

Feature Scaling is a preprocessing technique used to bring numerical features to a similar scale so that features with larger numerical values do not dominate the model. Common techniques are Standardization and Min-Max Normalization.
