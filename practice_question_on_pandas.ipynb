```python
#Count the number of employees in each department

#Best when you only need the frequency of values in one column.

print(df["Department"].value_counts())
```

```python
# groupby().size() → Better when you're already using groupby() or need more complex grouped calculations.
print(df.groupby("Department").size().reset_index(name="Employee_Count"))
```
